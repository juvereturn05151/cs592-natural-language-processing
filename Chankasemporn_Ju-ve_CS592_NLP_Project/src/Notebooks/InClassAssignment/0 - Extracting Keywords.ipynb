1.1 Extracting keywords from a document
====

In this demo, we will explore the various ways of extracting keywords from a file using the examples of nltk's stemmers and lemmatizers.



```python
from pathlib import Path

# Add the project root (two levels up from this notebook) to the system path
# so that 'src' can be recognized as a module.
import os, sys
project_root = os.path.abspath(os.path.join(os.getcwd(), "..", ".."))
if project_root not in sys.path:
    sys.path.append(project_root)

from src.NLP_Globals import trainDir
import xml.etree.ElementTree as ET

cwd = Path.cwd()

src_dir = cwd
while src_dir.name != "src":
    src_dir = src_dir.parent

repo_root = src_dir.parent
sys.path.insert(0, str(src_dir))

train_dir = repo_root / "data" / "train"
file_path = train_dir / "Conan_Doyle_1.txt"
tree = ET.parse(file_path)

queryStr = "Elementary! My dear Watson!"

print(tree)

root = tree.getroot()

bodyNode = root.findall('Body')[0]
itemList = bodyNode.findall('.//Item')
print( f"Found {len(itemList)} elements")
# print(itemList)
```

Now go through each element and gather the tokens

```python
# Next step - find the words that occur most frequently in this document
def  buildWordMap(documentTokens):
    wordMap = dict()
    tempMap = {}
    for word in documentTokens:
        if word in tempMap:
            tempMap[word] += 1
        else:
            if word != "":
                tempMap[word] = 1
    # return tempMap
            
    numTokens = len(documentTokens)
    for word in tempMap:
        count = tempMap[word]
        wordMap[word] = (float(float(count) / float(numTokens)).__format__("1.4f"), count)

    # Sort the map based on the values
    wordMap = dict(sorted(wordMap.items(), key=lambda x: x[1], reverse=True))
    
    print(f"{len(wordMap)} distinct words found")
    return wordMap

# This is a hack to print the words from the dictionary into a sorted order
def  printSortedMap(inputMap):
    dummyList = []
    for key, value in inputMap.items():
        dummyList.append([value, key])
    dummyList = sorted(dummyList, reverse=True)
    return dummyList
```

```python
import src.InClassResource.TokenizerHelper as th

# This is a helper class that encapsulates various tokenization mechanisms

docTokens = []
for child in itemList:
    print("Using built-in split method")
    # 1. Using the built-in tokenizer from Python string, using 'split()' method
    # tokenlist = textBlock.split()
    docTokens += th.getTokensSplit(child)
    
print(docTokens)
```
