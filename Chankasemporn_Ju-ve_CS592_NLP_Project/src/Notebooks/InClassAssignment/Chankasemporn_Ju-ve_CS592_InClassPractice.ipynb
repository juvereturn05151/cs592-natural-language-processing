1.Import Everything

```python
import sys
import xml.etree.ElementTree as ET

from pathlib import Path

cwd = Path.cwd()

src_dir = cwd
while src_dir.name != "src":
    src_dir = src_dir.parent

repo_root = src_dir.parent
sys.path.insert(0, str(src_dir))

train_dir = repo_root / "data" / "train"
file_path = train_dir / "Jules_Verne_Twenty_Thousand_Leagues.txt"  # must be XML

print("exists?", file_path.exists())
print(file_path)

tree = ET.parse(file_path)
root = tree.getroot()

body = root.find("Body")
if body is None:
    raise ValueError("No <Body> element found in this XML.")

item_list = body.findall(".//Item")
for item in item_list[:5]:
    print(item.tag, item.attrib)
    print(item.text)
```

2.Extract information on from the document
=

- From the document, extract the raw text contained within the '<' Item '>' '</'Item'>'tags
- Store the text in a suitable container object
- Print the text to console one by one by itera ng over the container


```python
# Find the Body element
bodyElement = root.find('Body')

itemList = bodyElement.findall(".//Item")
for item in itemList:
    print( item.tag )
    print( item.attrib )
    print( item.text )
```

3.getTokensSplit
=



```python
import src.InClassResource.TokenizerHelper

TH = src.InClassResource.TokenizerHelper.TokenizerHelper()

allTokens = []

for item in itemList:
    tokens = TH.getTokensSplit(item)
    allTokens.extend(tokens)

wordMap = TH.buildWordMap(allTokens)
TH.printSortedMap(wordMap)
```

4.getTokens_NLTK_Tokenize

5.Stemming
=



```python
allStemTokens = []

for item in itemList:
    stemTokens = TH.getTokens_NLTK_Stemmer(item, choice=0)  # 0=Porter, 1=Lancaster, 2=Lemma, 3=Snowball
    allStemTokens.extend(stemTokens)

stemWordMap = TH.buildWordMap(allStemTokens)
TH.printSortedMap(stemWordMap)
```
