```python
import string
import xml.etree.ElementTree as ET
import re
import nltk

FILE_LIST = [
    'Conan_Doyle_1.txt',
    ]

STOP_WORDS = nltk.corpus.stopwords.words('english')
VALID_TAGS = ['NN', 'NNP', 'NNPS', 'NNS', 'CD', 'FW', 'JJ', 'JJR', 'JJS']
REGEX_CLEANR = re.compile(r"['—_\“\”\"\”’‘\-)\:!\&]")
```

```python
def extractNormalizedFrequency(filePath):
    tree = ET.parse(filePath)
    root = tree.getroot()

    big_string = ""

    bodyNode = root.find('Body')

    itemList = bodyNode.findall('.//Item') # type: ignore

    for item in itemList:
        text = item.text
        if text is None:
            continue
        big_string += text + " "

    token_list = [re.sub(REGEX_CLEANR, "", word) for word in
    nltk.WordPunctTokenizer().tokenize(big_string) if word not in STOP_WORDS and 
    word not in string.punctuation]

    nltk_pos_tags = nltk.pos_tag(token_list)

    filtered_tokens = []
    for token, tag in nltk_pos_tags:
        if tag in VALID_TAGS:
            if token not in STOP_WORDS and token not in string.punctuation:
                filtered_tokens.append(token)

    token_counts = {}

    token_num = len(filtered_tokens)
    for token in filtered_tokens:
        if token in token_counts:
            token_counts[token] += 1
        else:
            token_counts[token] = 1

    normalized_frequencies = {}
    for token, count in token_counts.items():
        normalized_frequencies[token] = (count, count / token_num)

    sorted_frequencies = dict(sorted(normalized_frequencies.items(), key=lambda item: item[1][1], reverse=True))

    return sorted_frequencies
```
